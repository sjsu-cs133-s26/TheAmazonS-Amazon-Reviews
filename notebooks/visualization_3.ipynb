# Visualization 3: Rating Composition by Star Level

This notebook analyzes the distribution of ratings (1–5 stars) in the Amazon Books Reviews dataset.

The goal is to understand overall customer sentiment by examining how ratings are distributed.

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/Books_rating.csv")
df.head()
```

```python
rating_counts = df["review/score"].value_counts().sort_index()
rating_percent = (rating_counts / rating_counts.sum()) * 100

import matplotlib.pyplot as plt

plt.figure(figsize=(6,4))
rating_percent.plot(kind="bar")

plt.title("Rating Composition (1–5 Stars)")
plt.xlabel("Star Rating")
plt.ylabel("Percentage of Reviews")

plt.tight_layout()
plt.savefig("rating_composition.png")
plt.show()
```
