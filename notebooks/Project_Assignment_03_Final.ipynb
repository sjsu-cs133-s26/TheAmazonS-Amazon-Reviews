```python
import pandas as pd

reviews_path = "/content/amazon_books_data/Books_rating.csv"

df = pd.read_csv(reviews_path)

print(df.shape)
df.head()
```

```python
df.columns
```

```python
top_books = (
    df.dropna(subset=["Title"])
      .groupby("Title")
      .size()
      .reset_index(name="review_count")
      .sort_values(by="review_count", ascending=False)
      .head(10)
)

top_books
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))

sns.barplot(
    data=top_books,
    x="review_count",
    y="Title"
)

plt.title("Top 10 Most-Reviewed Books on Amazon")
plt.xlabel("Number of Reviews")
plt.ylabel("Book Title")
plt.tight_layout()

plt.show()
```
